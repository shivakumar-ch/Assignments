### Question 4
##### We can merge two dicts in many ways
- We can merge them using dictionary unpacking.
- We can copy and update to merge two dicts.

```python
#using unpacking
d1={'a':1, 'b':2}
d2={'c':3, 'd':4}
merged = {**d1, **d2}
print(merged)

#Using update method
d1={'a':1, 'b':2}
d2={'c':3, 'd':4}
d1.update(d2)
print(d1)
```

### Question 8

```python
words='stephen is working to gain experience but experience comes with working for many years, stephen knows it but not the value of it'.lower().split()
freq = {}
for word in words:
    freq[word] = freq.get(word, 0) + 1
print(freq)
```

### Question 10
##### Sort a dictionary by keys and values.

```python
scores = {"charlie": 88, "alice": 95, "bob": 95, "dave": 72}

# Sort by key 
by_key = dict(sorted(scores.items(), key=lambda item: item[0]))
print(by_key)

# Sort by value 
by_value_asc = dict(sorted(scores.items(), key=lambda item: (item[1], item[0])))
print(by_value_asc)

# Sort by value 
by_value_desc = dict(sorted(scores.items(), key=lambda item: item[1], reverse=True))
print(by_value_desc)
```

### Question 13

Grouping lists of anagrams

```python

words = ["listen", "silent", "enlist", "hello", "lemon", "melon", "debitcard", "badcredit"]

groups = {}
for word in words:
    canonical_key = "".join(sorted(word))
    if canonical_key not in groups:
        groups[canonical_key] = []
    groups[canonical_key].append(word)

result = list(groups.values())
print(result)
```

### Question 14

##### Merging two dictionary by adding common keys.


```python
d1 = {"a": 10, "b": 20, "c": 30}
d2 = {"b": 15, "c": 5, "d": 40}

merged = d1.copy()
for key, value in d2.items():
    merged[key] = merged.get(key, 0) + value

print(merged)
```

### Question 16
##### Mirror Map

```python
mirror_map = {
    **{
        chr(ord("A") + i): chr(ord("Z") - i)
        for i in range(26)
    },
    **{
        chr(ord("a") + i): chr(ord("z") - i)
        for i in range(26)
    },
}

text = "Hello World"
mirrored = "".join(mirror_map.get(ch, ch) for ch in text)

print(mirrored)
```

### Question 26

Find Min and Max numbers in a set.

```python
numbers = {42, 7, -24, -3, 18}

if not numbers:
    raise ValueError("Cannot find min/max of an empty set")

iterator = iter(numbers)
first = next(iterator)
minimum = maximum = first

for item in iterator:
    if item < minimum:
        minimum = item
    if item > maximum:
        maximum = item

print(f"Min: {minimum}, Max: {maximum}")  
```

### Question 28

Intersection of two sets with & without built-in method.

```python
s1 = {1, 2, 3, 4, 5}
s2 = {3, 4, 5, 6, 7}

#Using built-in method
with_method = s1.intersection(s2) 
print(with_method)  

#Without built-in methods 
without_method = set()
for item in s1:
    if item in s2:
        without_method.add(item)
print(without_method)
```

### Question 29

Find the symmetric difference of two sets with and without using built-in methods.

```python
s1 = {1, 2, 3, 4}
s2 = {3, 4, 5, 6}

# Using built-in method 
with_method = s1.symmetric_difference(s2) 
print(with_method)

# Without built-in methods 
without_method = set()
for item in s1:
    if item not in s2:
        without_method.add(item)
for item in s2:
    if item not in s1:
        without_method.add(item)
print(without_method)
```

### Question 30

Set Operations.

```python
set1 = {1, 2, "apple", "orange"}
set2 = {2, 3, "orange", "banana"}

# Union 
union_result = set()
for item in set1:
    union_result.add(item)
for item in set2:
    union_result.add(item)

# Intersection 
intersection_result = set()
for item in set1:
    if item in set2:
        intersection_result.add(item)

# Difference: set1 - set2
diff_1_minus_2 = set()
for item in set1:
    if item not in set2:
        diff_1_minus_2.add(item)

# Difference: set2 - set1 
diff_2_minus_1 = set()
for item in set2:
    if item not in set1:
        diff_2_minus_1.add(item)

print("Union:", union_result)               
print("Intersection:", intersection_result) 
print("set1 - set2:", diff_1_minus_2)      
print("set2 - set1:", diff_2_minus_1)       
```
